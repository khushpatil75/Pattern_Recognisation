```python
# 1️⃣ Import Libraries
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
from ipywidgets import interact
import json
from io import BytesIO
from IPython.display import display
from ipycanvas import Canvas

IMG_SIZE = 128
BATCH = 32
EPOCHS = 1

# ---------------- Load Data ---------------- #
train_dir = r"data/devanagari/train"
test_dir  = r"data/devanagari/test"

datagen = ImageDataGenerator(rescale=1/255.)
train_ds = datagen.flow_from_directory(train_dir,
                                       target_size=(IMG_SIZE, IMG_SIZE),
                                       batch_size=BATCH,
                                       class_mode='categorical')
test_ds = datagen.flow_from_directory(test_dir,
                                      target_size=(IMG_SIZE, IMG_SIZE),
                                      batch_size=BATCH,
                                      class_mode='categorical')

# Save label mappings
labels_map = {v: k for k, v in train_ds.class_indices.items()}
with open("labels_map.json", "w") as f:
    json.dump(train_ds.class_indices, f)
print("Labels saved!")

# ---------------- Train CNN ---------------- #
cnn = models.Sequential([
    layers.Input(shape=(IMG_SIZE, IMG_SIZE, 3)),
    layers.Conv2D(32, 3, activation="relu"),
    layers.MaxPooling2D(),
    layers.Conv2D(64, 3, activation="relu"),
    layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dense(len(labels_map), activation="softmax")
])
cnn.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])
cnn.fit(train_ds, validation_data=test_ds, epochs=EPOCHS)
cnn.save("models/cnn.keras")
print("CNN Saved!")

# ---------------- Load Models ---------------- #
cnn = tf.keras.models.load_model("models/cnn.keras")
vit = tf.keras.models.load_model("models/vit.keras")      # load after training
hybrid = tf.keras.models.load_model("models/hybrid.keras")

models_dict = {"CNN": cnn, "ViT": vit, "Hybrid": hybrid}

# ---------------- Interactive Drawing Canvas ---------------- #
canvas = Canvas(width=256, height=256)
display(canvas)

def predict_from_canvas():
    img = Image.fromarray(canvas.get_image_data("uint8")).convert("RGB")
    img = img.resize((IMG_SIZE, IMG_SIZE))
    img = np.array(img)/255.0
    img = img.reshape(1, IMG_SIZE, IMG_SIZE, 3)
    
    results = {}
    for name, model in models_dict.items():
        pred = model.predict(img, verbose=0)[0]
        cls = np.argmax(pred)
        results[name] = (labels_map[cls], pred[cls]*100)
    
    for name, (label, conf) in results.items():
        print(f"{name}: {label} ({conf:.2f}%)")
    
interact(predict_from_canvas)
```
